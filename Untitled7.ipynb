# Ant Colony Optimization (ACO) for Network Packet Routing

This notebook implements an Ant Colony Optimization algorithm to find the optimal path for packet routing in a network. We will define the routing environment, implement the ACO routing algorithm (with pheromone deposition and evaporation), and analyze performance scalability as $N$ increases.

```python
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
import random

# Set random seed for reproducibility
np.random.seed(42)
random.seed(42)
```

## 1. Define the Network and Pheromone Functions

In network packet routing, we define:
- **Heuristic Information ($̧_{ij}$)**: Typically inversely proportional to link cost (latency, congestion, or distance). $̧_{ij} = 1 / d_{ij}$.
- **Pheromone Level ($̔_{ij}$)**: Represents the learned quality of link $(i, j)$.
- **Transition Probability**:
  $$P_{ij}^k = \frac{[̔_{ij}]^̑ [̧_{ij}]^̒}{\sum_{l \in \text{allowed}} [̔_{il}]^̑ [̧_{il}]^̒}$$
- **Pheromone Update**:
  $$̔_{ij} \leftarrow (1 - ̡)̔_{ij} + \sum_{k} \Delta ̔_{ij}^k$$
  where $̡$ is the evaporation rate, and $\Delta ̔_{ij}^k = Q / L_k$ if ant $k$ used edge $(i, j)$ (with $L_k$ being the total path cost).

```python
class NetworkRoutingACO:
    def __init__(self, num_nodes, alpha=1.0, beta=2.0, evaporation_rate=0.1, Q=10.0):
        self.N = num_nodes
        self.alpha = alpha  # Pheromone importance
        self.beta = beta    # Heuristic importance
        self.rho = evaporation_rate
        self.Q = Q

        # Generate random coordinates for network nodes
        self.node_positions = {i: (np.random.rand() * 100, np.random.rand() * 100) for i in range(num_nodes)}

        # Initialize distance matrix (fully connected for simplicity, high values for no-link)
        self.distances = np.zeros((num_nodes, num_nodes))
        for i in range(num_nodes):
            for j in range(num_nodes):
                if i == j:
                    self.distances[i][j] = np.inf
                else:
                    pos_i = self.node_positions[i]
                    pos_j = self.node_positions[j]
                    self.distances[i][j] = np.sqrt((pos_i[0] - pos_j[0])**2 + (pos_i[1] - pos_j[1])**2)

        # Initialize pheromones
        self.pheromones = np.ones((num_nodes, num_nodes)) * 0.1

    def get_heuristics(self):
        # Heuristic is 1/distance
        with np.errstate(divide='ignore'):
            heuristics = 1.0 / self.distances
        heuristics[np.isinf(heuristics)] = 0
        return heuristics

    def transition_probabilities(self, current_node, visited):
        pheromones = self.pheromones[current_node]
        heuristics = self.get_heuristics()[current_node]

        # Mask visited nodes
        mask = np.ones(self.N, dtype=bool)
        mask[list(visited)] = False

        numerator = (pheromones ** self.alpha) * (heuristics ** self.beta)
        numerator[~mask] = 0

        sum_numerator = np.sum(numerator)
        if sum_numerator == 0:
            # If stuck, distribute probability equally among unvisited
            if np.any(mask):
                return mask.astype(float) / np.sum(mask)
            return np.zeros(self.N)

        return numerator / sum_numerator

    def update_pheromones(self, paths, path_costs):
        # Evaporation
        self.pheromones *= (1 - self.rho)
        # Ensure a minimum pheromone level to prevent total stagnation
        self.pheromones = np.maximum(self.pheromones, 1e-4)

        # Deposition
        for path, cost in zip(paths, path_costs):
            if cost > 0 and cost != np.inf:
                deposit = self.Q / cost
                for i in range(len(path) - 1):
                    u, v = path[i], path[i+1]
                    self.pheromones[u][v] += deposit
                    self.pheromones[v][u] += deposit # Undirected routing graph
```

## 2. Implement Routing Optimization Run

```python
def find_optimal_route(aco, source, destination, num_ants=15, generations=50):
    best_path = None
    best_cost = np.inf

    for gen in range(generations):
        paths = []
        path_costs = []

        for ant in range(num_ants):
            path = [source]
            visited = {source}
            current = source

            while current != destination:
                probs = aco.transition_probabilities(current, visited)
                if np.sum(probs) == 0:
                    # Backtrack or terminate path if dead end
                    break

                next_node = np.random.choice(aco.N, p=probs)
                path.append(next_node)
                visited.add(next_node)
                current = next_node

            if path[-1] == destination:
                # Calculate path cost
                cost = sum(aco.distances[path[i]][path[i+1]] for i in range(len(path) - 1))
                paths.append(path)
                path_costs.append(cost)

                if cost < best_cost:
                    best_cost = cost
                    best_path = path

        # Update pheromones based on the performance of the paths discovered in this generation
        if paths:
            aco.update_pheromones(paths, path_costs)

    return best_path, best_cost

# Run simulation
N_nodes = 15
aco_instance = NetworkRoutingACO(num_nodes=N_nodes)
source_node, dest_node = 0, N_nodes - 1
path, cost = find_optimal_route(aco_instance, source_node, dest_node)

print(f"Best Path found: {path}")
print(f"Path Cost (Latency/Distance Metric): {cost:.2f}")
```

```python
import matplotlib.pyplot as plt
import networkx as nx

# Create a NetworkX Graph from the simulation setup to show our input network
G = nx.Graph()
for i in range(aco_instance.N):
    G.add_node(i, pos=aco_instance.node_positions[i])

# Fully connect the graph with weights corresponding to Euclidean distances
for i in range(aco_instance.N):
    for j in range(i + 1, aco_instance.N):
        dist = aco_instance.distances[i][j]
        G.add_edge(i, j, weight=round(dist, 1))

# Plot the input network topology
plt.figure(figsize=(10, 8))
pos = nx.get_node_attributes(G, 'pos')

# Draw nodes
nx.draw_networkx_nodes(G, pos, node_size=600, node_color='lightblue', edgecolors='black')
# Draw node labels
nx.draw_networkx_labels(G, pos, font_size=12, font_family='sans-serif')

# Highlight the optimal route discovered in cell 43b8d670
if 'path' in locals() and path:
    path_edges = list(zip(path[:-1], path[1:]))
    # Draw background grid/all edges lightly
    nx.draw_networkx_edges(G, pos, alpha=0.1, edge_color='gray')
    # Draw path edges prominently
    nx.draw_networkx_edges(G, pos, edgelist=path_edges, edge_color='red', width=3)
    plt.title(f'Input Network Topology (Size N={aco_instance.N}) with Solved ACO Route in Red')
else:
    nx.draw_networkx_edges(G, pos, alpha=0.2, edge_color='gray')
    plt.title(f'Input Network Topology (Size N={aco_instance.N})')

plt.grid(True)
plt.show()
```

## 3. Scale and Diversification Analysis ($N$ Scaling)

### What happens when $N$ is increased and diversified?
1. **State Space Explosion**: The search space of potential routes grows factorially ($O(N!)$). As a result, finding the global optimum requires significantly more ants and generations.
2. **Pheromone Dilution & Stagnation**: With more alternative links, pheromone trails can get dispersed (diluted) across many sub-optimal paths. Conversely, ants might converge too early on a poor path due to lack of exploration.
3. **Diversification Necessity**: To counteract scaling issues, we must increase exploration ($̒$ or dynamic modification of evaporation parameter $̡$). Let's visualize the computational scaling below.

```python
import time

node_sizes = [10, 20, 30, 40, 50]
times_taken = []
best_costs = []

for n in node_sizes:
    start = time.time()
    aco_scale = NetworkRoutingACO(num_nodes=n)
    _, b_cost = find_optimal_route(aco_scale, source=0, destination=n-1, num_ants=10, generations=20)
    times_taken.append(time.time() - start)
    best_costs.append(b_cost)

fig, ax1 = plt.subplots(figsize=(10, 5))

color = 'tab:blue'
ax1.set_xlabel('Network Size (N)')
ax1.set_ylabel('Execution Time (seconds)', color=color)
ax1.plot(node_sizes, times_taken, marker='o', color=color, linewidth=2)
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = 'tab:red'
ax2.set_ylabel('Optimal Path Cost Found', color=color)
ax2.plot(node_sizes, best_costs, marker='s', linestyle='--', color=color, linewidth=2)
ax2.tick_params(axis='y', labelcolor=color)

plt.title('ACO Routing Performance: Scaling with Node Count (N)')
fig.tight_layout()
plt.show()
```
